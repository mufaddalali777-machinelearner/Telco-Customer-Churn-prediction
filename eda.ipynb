The Telco customer churn dataset contains 7043 rows and 21 columns. Each columun in the dataset helps us in identifying the churn status of the customer.


Column 1: The unique customer id of each customer.

Column 2: Gender of the customer (Male or Female).

Column 3: Wether the customer is a Senior Citizen or not.

Column 4: Wether the customer has a patner or not.

Column 5: Wether the customer has dependents or not.

Column 6: Number of months the customer stayed with the company.

Column 7: Wether the customer had Phone Services or not.

Column 8: Wether the customer had multiple lines or not.

Column 9: The type of internet service selected by the customer.

Column 10-15: Types of internet services or addons selected by the customer.

Column 16: Contract period during which the customer took the benefit of the telecom services.

Column 17-20: Payment related information of the customer.

Column 21: Tells us wether the customer continues to use the services or not.

```python
# total charges had empty spaces

df["TotalCharges"] = df["TotalCharges"].replace(" ", None).astype(float)
```

```python
df["Contract"].value_counts().plot(kind = "bar")
plt.title("Contract periods of various customers")
plt.show()
```

Most customers prefer monthly contract.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df["TotalCharges"], bins=30)
plt.xlabel("Total Charges")
plt.ylabel("Frequency")
plt.title("Distribution of Total Charges")
plt.show()
```

most customers pay charges less than 2000$
