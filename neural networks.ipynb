```python
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.activations import linear, relu
from tensorflow.keras.layers import Dense, LeakyReLU
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")
df.head()
```

```python
df_model = df.copy()
df_model.head()
```

```python
df_model["Churn"] = df_model["Churn"].map({"Yes":1, "No":0})
df_model.head()
```

```python
df_model.drop("customerID", axis = 1, inplace = True)
df_model.head()
```

```python
df_model = pd.get_dummies(df_model, drop_first = True)
df_model.head()
```

```python
x = df_model.drop("Churn", axis = 1)
y = df_model["Churn"]
```

```python
x_train, x_, y_train, y_ = train_test_split(x, y, test_size = 0.3, random_state = 1)
x_cv, x_test, y_cv, y_test = train_test_split(x_, y_, test_size = 0.5, random_state = 1)
```

```python
linear_scaler = StandardScaler()
x_train_sc = linear_scaler.fit_transform(x_train)
x_cv_sc = linear_scaler.transform(x_cv)
x_test_sc = linear_scaler.transform(x_test)

x_train_sc
```

```python
def nn_models():
    tf.random.set_seed(42)

    model_1 = Sequential([
        
        Dense(16, activation= "relu", kernel_regularizer = tf.keras.regularizers.l2(0.005)),
        Dense(8, activation="relu", kernel_regularizer = tf.keras.regularizers.l2(0.005)),
        Dense(1, activation= "sigmoid",kernel_regularizer = tf.keras.regularizers.l2(0.005))
    ],name = 'model_1')

    model_2 = Sequential([
        Dense(6, activation= "relu", kernel_regularizer = tf.keras.regularizers.l2(0.005)),
        Dense(4, activation= "relu", kernel_regularizer = tf.keras.regularizers.l2(0.005)),
        Dense(3, activation="relu", kernel_regularizer = tf.keras.regularizers.l2(0.005)),
        Dense(1, activation= "sigmoid",kernel_regularizer = tf.keras.regularizers.l2(0.005))
    ], name = 'model_2')

    models = [model_1, model_2]

    return models
```

```python
from sklearn.metrics import recall_score

nn_train_mse = []
nn_cv_mse = []
nn_cv_recall = []

for model in my_models:
    model.compile(
        loss='binary_crossentropy',
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001)
    )
    
    print(f"Training : {model.name}")
    
    model.fit(
        x_train_sc, y_train,
        epochs=20,
        verbose=1
    )
    
    print("Done!\n")

    # ---- MSE (your existing code) ----
    y_hat = model.predict(x_train_sc)
    nn_train_mse.append(mean_squared_error(y_train, y_hat) / 2)

    y_hat = model.predict(x_cv_sc)
    nn_cv_mse.append(mean_squared_error(y_cv, y_hat) / 2)

    # ---- NEW: Recall calculation ----
    y_pred = (y_hat >= 0.5).astype(int)
    recall = recall_score(y_cv, y_pred)
    print(classification_report(y_cv, y_pred))
    nn_cv_recall.append(recall)

print("Results:\n")
for i in range(len(nn_train_mse)):
    print(
        f"Model {i+1} : "
        f"Training MSE: {nn_train_mse[i]:.5f}, "
        f"CV MSE: {nn_cv_mse[i]:.5f}, "
        f"CV Recall (Churn=1): {nn_cv_recall[i]:.5f}"
    )
```
